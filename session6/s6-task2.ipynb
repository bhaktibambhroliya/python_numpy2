```python
import pandas as pd
```

```python
data = {
    "product_id": [101, 102, 101, 103, 102, 101],
    "user_id": [1, 2, 1, 3, 2, 4],
    "rating": [5, 4, 5, 3, 4, 2],
    "review_text": ["Good", "Nice", "Good", "Average",
                    "Nice", "Okay"]
}

df = pd.DataFrame(data)

before = df.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("Duplicate reviews before:", before)

df_cleaned = df.drop_duplicates(
    subset=["product_id", "user_id"],
    keep="first"
)

after = df_cleaned.duplicated(
    subset=["product_id", "user_id"]
).sum()

```

```python
print("Duplicate reviews after:", after)

print("Cleaned DataFrame:")
print(df_cleaned)
```
