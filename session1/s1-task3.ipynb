```python
import pandas as pd
```

```python
followers = pd.Series([1200, 54000, 780, 150000, 32000])

print("Followers:", followers)

highest = followers.idxmax()
lowest = followers.idxmin()

print("Highest Followers:")
print("Index:", highest)
print("Followers:", followers[highest])

print("Lowest Followers:")
print("Index:", lowest)
print("Followers:", followers[lowest])
```
