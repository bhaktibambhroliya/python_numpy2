```python
import pandas as pd
```

```python
data = {
    "Item": ["Pizza", "Burger", "Sandwich", "Pasta", "Fries"],
    "Price": [200, 120, 100, 180, 80],
    "Qty": [2, 3, 1, 2, 4]
}

df = pd.DataFrame(data)

df["DiscountedPrice"] = df["Price"].apply(
    lambda price: price * 0.90
)

```
