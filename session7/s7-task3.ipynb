```python

import pandas as pd

data = {
    "PostID": [101, 102, 103, 104, 105],
    "Likes": [1500, 2000000, 850, 12000, 500000],
    "Comments": [120, 2500, 50, 1500000, 800]
}

df = pd.DataFrame(data)

def format_number(num):
    if num >= 1000000:
        return f"{num / 1000000:g}M"
    elif num >= 1000:
        return f"{num / 1000:g}K"
    else:
        return str(num)

df[["Likes", "Comments"]] = df[["Likes", "Comments"]].map(
    format_number
)

print(df)
```
