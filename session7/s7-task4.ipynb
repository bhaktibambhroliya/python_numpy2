```python
import pandas as pd
```

```python
data = {
    "Restaurant": ["Dominos", "McDonalds", "Subway", "Pizza Hut", "KFC"],
    "Price": [500, 300, 250, 600, 400],
    "Qty": [2, 3, 2, 1, 2],
    "DeliveryCharge": [40, 30, 25, 50, 35]
}

df = pd.DataFrame(data)

df["FinalAmount"] = df.apply(
    lambda row: (row["Price"] * row["Qty"]) + row["DeliveryCharge"],
    axis=1
)

```
