```python
import pandas as pd




```

```python
df = pd.DataFrame({
    "name": [
        "Restaurant A",
        "Restaurant B",
        "Restaurant C",
        "Restaurant D",
        "Restaurant E"
    ],
    "rating": [4.5, 4.2, 3.8, 4.7, 4.0],
    "votes": [1200, 900, 400, 1500, 650],
    "avg_cost": [500, 400, 300, 700, 350]
})
```

```python
correlation = df["rating"].corr(df["votes"])
```

```python
print(df)
print("\nCorrelation between rating and votes:", round(correlation, 2))
```
