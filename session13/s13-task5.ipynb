```python
import pandas as pd

df = pd.read_csv("IPL_2023_Batsmen.csv")

# Column names check karein
print(df.columns)

df.columns = df.columns.str.strip()

if "runs" in df.columns:
    df["runs"] = pd.to_numeric(df["runs"], errors="coerce")
    print("Top 5 players by runs:")
    print(df.nlargest(5, "runs")[["player_name", "runs"]])
else:
    print("Runs column nahi mila. Actual column names:", df.columns.tolist())
```
