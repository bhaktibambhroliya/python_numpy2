```python
import pandas as pd



```

```python
df = pd.DataFrame({
    "user_id": [101, 102, 103, 104, 105],
    "start_date": [
        "01-01-2026",
        "15-03-2026",
        "01-06-2026",
        "10-08-2026",
        "01-10-2026"
    ]
})

df["start_date"] = pd.to_datetime(
    df["start_date"],
    format="%d-%m-%Y"
)

today = pd.Timestamp.today().normalize()

df["tenure_days"] = (today - df["start_date"]).dt.days
```
