```python
import pandas as pd



```

```python
df = pd.DataFrame({
    "movie": ["Movie A", "Movie B", "Movie C", "Movie D"],
    "release_date": [
        "15-01-2024",
        "20-06-2023",
        "10-12-2025",
        "05-03-2022"
    ]
})

# Convert release_date to datetime
df["release_date"] = pd.to_datetime(
    df["release_date"],
    format="%d-%m-%Y"
)

df["year"] = df["release_date"].dt.year
df["month"] = df["release_date"].dt.month
df["day"] = df["release_date"].dt.day
```
