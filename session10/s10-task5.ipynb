```python
import pandas as pd
```

```python
products = pd.DataFrame({
    "product_id": [101, 102, 103, 104],
    "name": ["Mobile", "Laptop", "Headphones", "Watch"],
    "category": ["Electronics", "Electronics", "Accessories", "Accessories"]
})
```

```python
product_ratings = pd.DataFrame({
    "rating_id": [1, 2, 3, 4],
    "product_id": [101, 102, 103, 105],
    "rating": [5, 4, 3, 5]
})
```

```python
result = pd.merge(
    products,
    product_ratings,
    on="product_id",
    how="right"
)
```

```python
print(result)
```
