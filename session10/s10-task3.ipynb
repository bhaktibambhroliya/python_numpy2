```python
import pandas as pd
```

```python
restaurants = pd.DataFrame({
    "restaurant_id": [1, 2, 3],
    "name": ["Food Palace", "Pizza Corner", "Spice Hub"],
    "city": ["Ahmedabad", "Surat", "Rajkot"]
})
```

```python
zomato_reviews = pd.DataFrame({
    "review_id": [101, 102, 103],
    "restaurant_id": [1, 2, 4],
    "rating": [5, 4, 3],
    "reviewer_city": ["Ahmedabad", "Surat", "Mumbai"]
})
```

```python
result = pd.merge(
    restaurants,
    zomato_reviews,
    on="restaurant_id",
    how="outer"
)
```

```python
print(result)
```
