```python
import pandas as pd
```

```python
users = pd.DataFrame({
    "user_id": [1, 2, 3, 4, 5],
    "username": ["Amit", "Priya", "Rahul", "Neha", "Riya"],
    "city": ["Ahmedabad", "Surat", "Rajkot", "Vadodara", "Ahmedabad"]
})
```

```python
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "user_id": [1, 2, 1, 3, 5, 2, 3],
    "order_amount": [500, 800, 300, 1200, 600, 450, 900]
})
```

```python
result = pd.merge(users, orders, on="user_id", how="inner")
```
