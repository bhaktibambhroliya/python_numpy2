```python
import pandas as pd
```

```python
playlists = pd.DataFrame({
    "playlist_id": [101, 102, 103, 104, 105],
    "user_id": [1, 2, 3, 4, 5],
    "city": ["Ahmedabad", "Surat", "Rajkot", "Mumbai", "Ahmedabad"],
    "genre": ["Pop", "Rock", "Jazz", "Hip-Hop", "Classical"]
})
```

```python
user_profiles = pd.DataFrame({
    "user_id": [1, 2, 3, 4, 5],
    "username": ["Amit", "Priya", "Rahul", "Neha", "Riya"],
    "city": ["Ahmedabad", "Surat", "Surat", "Mumbai", "Vadodara"]
})
```

```python
result = pd.merge(
    playlists,
    user_profiles,
    on=["user_id", "city"],
    how="inner"
)
```
