```python
import pandas as pd
```

```python
data = {
    "user": [
        "Amit", "Amit", "Amit", "Priya", "Priya", "Priya",
        "Rahul", "Rahul", "Rahul"
    ],
    "day": [
        "Monday", "Tuesday", "Wednesday",
        "Monday", "Tuesday", "Wednesday",
        "Monday", "Tuesday", "Wednesday"
    ],
    "streams": [100, 150, 120, 200, 180, 220, 90, 110, 130]
}
```

```python
df = pd.DataFrame(data)
```

```python
result = df.pivot(
    index="user",
    columns="day",
    values="streams"
)
```
