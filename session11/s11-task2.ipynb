```python
import pandas as pd
```

```python
products = pd.DataFrame({
    "product_id": [101, 102, 103, 104],
    "product_name": ["Mobile", "Laptop", "Headphones", "Watch"]
})
```

```python
ratings = pd.DataFrame({
    "rating": [5, 4, 3, 5]
})
```

```python
result = pd.concat(
    [products, ratings],
    axis=1
)
```

```python
print(result)
```
