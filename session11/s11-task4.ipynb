```python
import pandas as pd
```

```python
data = {
    "restaurant": [
        "Dominos", "Dominos", "Dominos",
        "KFC", "KFC", "KFC",
        "Subway", "Subway", "Subway"
    ],
    "month": [
        "January", "January", "February",
        "January", "February", "February",
        "January", "February", "March"
    ],
    "total_amount": [500, 700, 600, 400, 450, 550, 300, 350, 400]
}
```

```python
df = pd.DataFrame(data)
```

```python
result = df.pivot_table(
    index="restaurant",
    columns="month",
    values="total_amount",
    aggfunc="mean"
)
```

```python
print(result)
```
