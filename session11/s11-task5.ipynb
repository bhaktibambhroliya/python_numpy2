```python
import pandas as pd

```

```python
jan = pd.DataFrame({
    "product": ["Shoes", "T-Shirt", "Jeans"],
    "sales": [5000, 3000, 4000]
})
jan.to_csv("sales_jan.csv", index=False)
```

```python
feb = pd.DataFrame({
    "product": ["Shoes", "T-Shirt", "Jeans"],
    "sales": [6000, 3500, 4500]
})
feb.to_csv("sales_feb.csv", index=False)
```

```python
mar = pd.DataFrame({
    "product": ["Shoes", "T-Shirt", "Jeans"],
    "sales": [7000, 4000, 5000]
})
mar.to_csv("sales_mar.csv", index=False)
```

```python
jan = pd.read_csv("sales_jan.csv")
feb = pd.read_csv("sales_feb.csv")
mar = pd.read_csv("sales_mar.csv")
```

```python
all_sales = pd.concat(
    [jan, feb, mar],
    ignore_index=True
)
```

```python
print("Combined Sales:")
print(all_sales)
```

```python
total_sales = all_sales.groupby("product")["sales"].sum()
```

```python
print("Total Sales Per Product:")
print(total_sales)
```
