```python
import pandas as pd
```

```python
expenses = pd.DataFrame({
    "Date": ["01-10-2026", "02-10-2026", "03-10-2026",
             "04-10-2026", "05-10-2026"],
    "Expense": ["Recharge", "Food", "Travel",
                "Shopping", "Internet"],
    "Amount": [299, 150, 80, 500, 399]
})

expenses.to_excel("mobile_expenses.xlsx", index=False)

print("Excel file created successfully!")
```

```python
df = pd.read_excel("mobile_expenses.xlsx")
```
