```python
import pandas as pd
```

```python
data = {
    "product_name": ["Mobile", "Laptop", "Headphones", "Watch", "Tablet"],
    "rating": [4, 5, None, 3, 4]
}

df = pd.DataFrame(data)

mean_rating = df["rating"].mean()

df["rating"] = df["rating"].fillna(mean_rating)

```
