```python
import pandas as pd
```

```python
data = {
    "team": [
        "CSK", "MI", "RCB", "CSK",
        "MI", "RCB", "GT", "GT"
    ],
    "runs": [180, 165, 200, 175, 190, 185, 170, 160]
}
```

```python
df = pd.DataFrame(data)
```

```python
df.to_csv("ipl_matches.csv", index=False)
```

```python
df = pd.read_csv("ipl_matches.csv")
```

```python
result = df.groupby("team")["runs"].sum().reset_index()
```
