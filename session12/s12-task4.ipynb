```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "region": ["West", "East", "West", "East", "North", "North"],
    "revenue": [50000, 30000, 70000, 40000, 25000, 35000]
})
```

```python
result = df.groupby("region").agg({
    "revenue": ["sum", "max", "min"]
})
```

```python
print(result)
```
