```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "city": [
        "Ahmedabad", "Ahmedabad", "Surat",
        "Surat", "Ahmedabad", "Surat"
    ],
    "restaurant": [
        "Dominos", "Dominos", "KFC",
        "KFC", "KFC", "Dominos"
    ],
    "order_amount": [500, 700, 400, 600, 300, 800]
})
```

```python
result = df.groupby(
    ["city", "restaurant"]
).agg(
    average_order_amount=("order_amount", "mean")
)
```

```python
print(result)
```
