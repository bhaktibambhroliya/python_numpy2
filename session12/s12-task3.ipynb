```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "category": [
        "Electronics", "Electronics", "Fashion",
        "Fashion", "Electronics", "Fashion"
    ],
    "region": [
        "West", "East", "West",
        "East", "West", "West"
    ],
    "sales": [50000, 30000, 20000, 15000, 40000, 25000],
    "units_sold": [50, 30, 100, 75, 40, 125]
})
```

```python
result = df.groupby(
    ["category", "region"]
).agg(
    total_sales=("sales", "sum"),
    average_units_sold=("units_sold", "mean")
).reset_index()
```

```python
print(result)
```
