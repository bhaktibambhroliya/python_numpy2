```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "artist": [
        "Artist A", "Artist B", "Artist C",
        "Artist D", "Artist E", "Artist F"
    ],
    "genre": [
        "Pop", "Rock", "Pop",
        "Hip-Hop", "Rock", "Pop"
    ],
    "streams": [100000, 80000, 150000, 120000, 60000, 200000]
})
```

```python
result = df.groupby("genre").agg(
    total_streams=("streams", "sum"),
    average_streams=("streams", "mean")
)
```

```python
result = result.sort_values(
    by="total_streams",
    ascending=False
)
```
